```python
import pandas as pd

# Supongamos que ya tienes cargados tus DataFrames: df1 y df2

# Convertir los nombres de las columnas a minúsculas en ambos DataFrames
scopus.columns = scopus.columns.str.lower()
wos.columns = wos.columns.str.lower()

# Definir un diccionario para renombrar columnas de df2 usando df1 como referencia.
# Se mapea cada nombre de df2 al nombre equivalente en df1.
rename_dict_df2 = {
    # Título del artículo
    "article title": "title",
    
    # Páginas: df1 tiene "page start" y "page end"
    "start page": "page start",
    "end page": "page end",
    
    # Referencias citadas
    "cited references": "references",
    
    # Año de publicación
    "publication year": "year",
    
    # Idioma
    "language": "language of original document",
    
    # Abreviatura del título de la fuente
    "journal iso abbreviation": "abbreviated source title",
    # (Si en df2 aparece "journal abbreviation", podrías mapearlo de igual forma)
    
    # Número de artículo
    "article number": "art. no.",
    
    # Financiamiento (se consolidan en una única columna de referencia)
    # "funding orgs": "funding details",
    # "funding text": "funding details",
    "funding name preferred": "funding details",
    
    # Identificadores de autor
    "researcher ids": "author(s) id",
    "orcids": "author(s) id",
    
    # Enlace o URL del artículo
    "doi link": "link",
    
    # Citaciones
    "times cited, all databases": "cited by",
    # "times cited, wos core": "cited by",
    
    # Nombre de la conferencia
    "conference title": "conference name"
}

# Renombramos las columnas de df2
df2_renamed = wos.rename(columns=rename_dict_df2)
# display(df2_renamed)

# Función para combinar columnas duplicadas resultantes del renombrado
def combine_duplicate_columns(df):
    # Para cada columna que aparezca más de una vez
    for col in set(df.columns):
        if list(df.columns).count(col) > 1:
            # Combina las columnas duplicadas: se toma el primer valor no nulo en cada fila
            df[col] = df.loc[:, df.columns == col].bfill(axis=1).iloc[:, 0]
            # Luego se eliminan las columnas duplicadas, conservando la primera
            df = df.loc[:, ~df.columns.duplicated(keep='first')]
    return df

df2_renamed = combine_duplicate_columns(df2_renamed)
display(df2_renamed)
```

## 3. Peer-review venues 	
The peer-review process is the evaluation of a research paper by experts in the same academic field prior to publication. This ensures that the content is rigorous, valid, original and relevant, helping to maintain the quality and integrity of the scientific literature.

Double-blind is still peer review. In fact, the double-blind review process-where neither the authors nor the reviewers know the identity of the other-is implemented to minimize bias. In addition, having at least three reviewers is usually a common practice to ensure a more complete and rigorous evaluation of the work.

In the single-blind review process, the authors do not know who the reviewers are, but the reviewers do know the identity of the authors. This allows the reviewers to evaluate the work taking into account the author's context, but it can also introduce biases. Despite this, it is still a widely used peer review process in the academic community.

```python
from pprint import pprint
import re

# 3. Peer-review venues 	
# Nota: analizar conferencias que faltan

def mark_peer_review(conference_name):
    """
    Recibe el nombre de una conferencia y devuelve True si se detecta que es
    una venue peer-reviewed, False en caso contrario.
    La función utiliza el diccionario peer_review_dict para buscar coincidencias
    parciales (ignorando mayúsculas y espacios adicionales).
    """
    # Diccionario original
    peer_review_dict = {
        # Conferencias principales y journals reconocidos
        "CVPR": True, # IEEE/CVF Conference: http://cvpr.thecvf.com/
        "ICML": True, # International Conference on Machine Learning: https://icml.cc/
        "NeurIPS": True, # Conference on Neural Information Processing Systems: https://neurips.cc/
        "ACL": True, # Annual Meeting of the ACL: https://www.aclweb.org/portal/
        "EMNLP": True, # Conference on Empirical Methods in NLP: https://aclanthology.org/venues/emnlp/
        "AAAI": True,  # AAAI Conference on AI: https://aaai.org/conference/
        "IJCAI": True, # International Joint Conference on AI: https://ijcai.org/
        "Nature": True,  # Nature Journal: https://www.nature.com/
        "PLoS ONE": True, # PLoS ONE Journal (for society): https://journals.plos.org/plosone/
        "LNCS": True,  # Lecture Notes in Computer Science (Springer). Springer Lecture Notes: https://link.springer.com/conference/lncs (?)
        "TOG": True,   # ACM Transactions on Graphics. ACM Transactions on Graphics: https://dl.acm.org/journal/tog
        "KDD": True,   # ACM SIGKDD. ACM SIGKDD Conference: https://www.kdd.org/
        
        # Conferencias de tu lista
        "NLPCC": False,   # CCF Conference (China), no es top-tier. CCF NLP Conference: http://tcci.ccf.org.cn/
        "ICEMCE": True,    # IEEE-affiliated (China). IEEE Conference: http://www.icemce.org/
        "ICAICE": True,    # Asumido peer-reviewed (conferencia internacional genérica - China ). International Conference on AI Eng.: http://www.icaice.org/
        "IJCNLP": True,    # Colaboración ACL-Asia. Joint ACL-Asia Conference: https://www.aclweb.org/portal/ijcnlp
        "CCL": True,       # Conferencia nacional china (bajo CCF). Chinese Comp. Ling. Conference: http://cips-cl.org/static/CCL2024/en/index.html

        "SIGIR": True,     # ACM SIGIR (top en Information Retrieval). ACM SIGIR Conference: https://sigir.org/
        "EEKE": True,      # Workshop bajo conferencias principales. Workshop @ JCDL/others: https://eeke-workshop.github.io/
        "COLING": True,    # Conferencia líder en lingüística computacional. Double-blind. International Conference on Comp. Ling.: https://coling2024.org/
        "NNICE": True,     # Asumido peer-reviewed (IEEE/Springer). IEEE Conference: http://www.nnice.org/ (?)
        "Big Data": True,  # IEEE Big Data Conference. IEEE Big Data Conference: http://bigdataieee.org/
        "CompleNet": True, # Conferencia internacional establecida. International Conference: https://complenet.weebly.com (?)
        "RANLP": True,     # Reconocida en NLP (Bulgaria). Recent Advances in NLP: https://ranlp.org/
        "MLNLP": True,     # Conferencia genérica (peer-reviewed típico). Machine Learning & NLP Conf.: https://ieeexplore.ieee.org/xpl/conhome/10799945/proceeding
        "ADMA": True,      # Springer LNCS proceedings. Advanced Data Mining Conf.: https://adma2023.org/
        
        # Plataformas y otros
        "arXiv": False,    # Preprints no revisados. Preprint Repository: https://arxiv.org/
        "NAACL": True,     # Capítulo norteamericano de ACL. NAACL Conference: https://naacl.org/
        "CIKM": True,      # ACM CIKM. ACM CIKM: https://cikm2024.org
        "DASFAA": True,    # Conferencia de bases de datos (Springer). Database Systems Conf.: https://www.dasfaa2024.org/
        "WWW": True,       # The Web Conference. The Web Conference: https://www2024.thewebconf.org/
        "LREC": True,      # Language Resources and Evaluation. Language Resources Conf.: https://lrec-coling-2024.org/
        "ECML": True,      # European Conference on Machine Learning. European Conf. on ML: https://2023.ecmlpkdd.org/
        "KSEM": True,      # Springer proceedings. Knowledge Science Conf.: https://ai-edge.net/index.html
        "IJCNN": True,     # IEEE International Joint Conference on Neural Networks. IEEE Neural Networks Conf.: https://2025.ijcnn.org
        "BIBM": True,      # IEEE Bioinformatics. IEEE Bioinformatics Conf.: https://ieeebibm.org/BIBM2024/
        "FLLM": True,      # Conferencia emergente (asumido peer-reviewed). Foundation & LLM Conf.: https://fllm-conference.org/ (?)
        "WSDM": True,      # ACM WSDM. ACM Web Search Conf.: https://www.wsdm-conference.org/
        "SMC": True,       # IEEE Systems, Man, and Cybernetics. IEEE Systems Conf.: https://ieeesmc2024.org/ 
        "ICICS": True      # IEEE/Springer conference. IEEE ICICS:https://icis2024.aisconferences.org 
}



    # Si no es un string (nan), se devuelve False.
    if not isinstance(conference_name, str):
        return "?"
   
    # Convertir el nombre de la conferencia a minúsculas
    conf_lower = conference_name.lower()
    conf_lower_list = conf_lower.split()
    # print(conf_lower)
    # Recorrer cada clave del diccionario
    for key, is_peer in peer_review_dict.items():
        # Convertir la clave a minúsculas para hacer la comparación
        key_lower = key.lower()
        print(conf_lower_list, key_lower)
        # Si la clave se encuentra en el nombre de la conferencia, devolver el valor asociado.
        if key_lower in conf_lower_list:
            return is_peer # True or false surely

    # Si no se encuentra ninguna coincidencia, se asume que no es peer reviewed
    return "?"
```
