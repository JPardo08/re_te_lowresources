```python
import os 
import pandas as pd

def load_all_datasets(main_path, queque, dataset):
    scopus = pd.DataFrame()
    que = os.listdir(main_path)
    for i in que:
        # Saltar elementos que terminen en .csv o .ipynb o sean ".DS_Store"
        if i.endswith('.csv') or i.endswith('.ipynb') or i == ".DS_Store" or i in queque:
            continue
        p = os.path.join(main_path, i)
        que2 = os.listdir(p)
        for a in que2:
            if a.endswith('.csv') or a.endswith('.ipynb') or a == ".DS_Store":
                continue
            dat = os.path.join(p, a)
            que3 = os.listdir(dat)
            if que3:
                ar = os.path.join(dat, dataset)
                if dataset.endswith(".xls"):
                    r = pd.read_excel(ar, engine="xlrd")
                    r["query"]=i
                else:
                    r = pd.read_csv(ar)
                    r["query"]=i
                scopus = pd.concat([scopus, r], ignore_index=True)
    return scopus
```

```python
# Ejemplo de uso:
main_path = "./" #/web_of_science
wos = load_all_datasets(main_path, queque=[], dataset="savedrecs.xls")
display(wos)
```

```python
import pandas as pd

# Supón que ya tienes el DataFrame 'df'
initial_rows = wos.shape[0]  # Número de filas inicial

# Eliminar filas duplicadas (puedes usar inplace=True para modificar el mismo DataFrame)
wos_final = wos.drop_duplicates(subset='Article Title', keep='first')

final_rows = wos_final.shape[0]  # Número de filas después de eliminar duplicados

filas_eliminadas = initial_rows - final_rows

print(f"Se han eliminado {filas_eliminadas} filas duplicadas.")
```

```python
ruta=os.path.join(main_path, "wos.csv")
wos_final.to_csv(ruta)
```
