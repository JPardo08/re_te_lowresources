```python
import pandas as pd

# Suponiendo que la columna "Description" contiene cadenas como:
# "Relation Extraction, Low-resource"
terms = df["Description"].tolist()

# Dividir cada descripción en dos términos y eliminar espacios extra
pairs = [tuple(map(str.strip, term.split(','))) for term in terms if ',' in term]

# Crear el DataFrame con dos columnas a partir de la lista de tuplas
terms_df = pd.DataFrame(pairs, columns=["term1", "term2"])

display(terms_df)
```

```python
from pybliometrics.scopus import ScopusSearch

# Hacer queries

for _, row in terms_df.iterrows():
    # print(row["term1"])
    query = f"TITLE-ABS-KEY('{row['term1'].lower()}' AND '{row['term2'].lower()}') AND PUBYEAR > 2015"
    print(f"Query: {query}")
    
    s = ScopusSearch(query)
    print(f"Number of results: {len(s.results)}")
```

```python
import os 
import pandas as pd

def load_all_datasets(main_path, queque, dataset):
    scopus = pd.DataFrame()
    que = os.listdir(main_path)
    for i in que:
        # Saltar elementos que terminen en .csv o .ipynb o sean ".DS_Store"
        if i.endswith('.csv') or i.endswith('.ipynb') or i == ".DS_Store" or i in queque:
            continue
        p = os.path.join(main_path, i)
        que2 = os.listdir(p)
        for a in que2:
            if a.endswith('.csv') or a.endswith('.ipynb') or a == ".DS_Store":
                continue
            dat = os.path.join(p, a)
            que3 = os.listdir(dat)
            if que3:
                ar = os.path.join(dat, dataset)
                if dataset.endswith(".xls"):
                    r = pd.read_excel(ar, engine="xlrd")
                    r["query"]=i
                else:
                    r = pd.read_csv(ar)
                    r["query"]=i
                scopus = pd.concat([scopus, r], ignore_index=True)
    return scopus
```

```python
# Ejemplo de uso:
main_path = "./" #/scopus
scopus = load_all_datasets(main_path, queque=[], dataset= "scopus.csv")
display(scopus)
```

```python
import pandas as pd

# Supón que ya tienes el DataFrame 'df'
initial_rows = scopus.shape[0]  # Número de filas inicial

# Eliminar filas duplicadas (puedes usar inplace=True para modificar el mismo DataFrame)
scopus_final = scopus.drop_duplicates()

final_rows = scopus_final.shape[0]  # Número de filas después de eliminar duplicados

filas_eliminadas = initial_rows - final_rows

print(f"Se han eliminado {filas_eliminadas} filas duplicadas.")
scopus_final.to_csv(os.path.join(main_path, "scopus.csv"))
```
